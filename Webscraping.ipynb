## Webscraping 

```python
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import time
import random

def scrape_adana_with_features():
    options = uc.ChromeOptions()
    driver = uc.Chrome(options=options)
    all_rows = []

    try:
        for page in range(1, 62):
            url = f"https://www.hepsiemlak.com/adana-kiralik?page={page}"
            driver.get(url)
            print(f"🚀 Processing Page {page}...")

            # 1. WAIT for the listing cards to actually exist
            try:
                WebDriverWait(driver, 15).until(
                    EC.presence_of_all_elements_located((By.CLASS_NAME, "listing-item"))
                )
            except:
                print(f"⚠️ Page {page} took too long to load. Skipping...")
                continue

            # 2. SCROLL slowly to trigger "Lazy Loading" (This fixes the N/A issue!)
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight/2);")
            time.sleep(2)
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
            time.sleep(2)

            # 3. Use JavaScript to grab the RAW text of every card
            # This is much safer than specific class names
            cards_text = driver.execute_script("""
                let data = [];
                let items = document.querySelectorAll('.listing-item');
                items.forEach(item => {
                    data.push(item.innerText.split('\\n').join(' | '));
                });
                return data;
            """)

            for text in cards_text:
                all_rows.append({"raw_text": text})
            
            # Save progress immediately so you don't lose data
            pd.DataFrame(all_rows).to_csv("adana_raw_v2.csv", index=False, encoding="utf-8-sig")
            print(f"✅ Total captured so far: {len(all_rows)}")

    finally:
        driver.quit()
        print("Done!")

scrape_adana_with_features()
```
